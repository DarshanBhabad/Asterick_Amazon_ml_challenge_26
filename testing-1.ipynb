```python
import pandas as pd

base = "/kaggle/input/datasets/bboyattitude/asterick-amazon-ml-ch-26/dataset"

train_s1 = pd.read_csv(f"{base}/train/train_source1.tsv", sep="\t", dtype=str)
train_s2 = pd.read_csv(f"{base}/train/train_source2.tsv", sep="\t", dtype=str)
train_s3 = pd.read_csv(f"{base}/train/train_source3.tsv", sep="\t", dtype=str)
gt        = pd.read_csv(f"{base}/train/train_ground_truth.tsv", sep="\t", dtype=str)

test_s1 = pd.read_csv(f"{base}/test/test_source1.tsv", sep="\t", dtype=str)
test_s2 = pd.read_csv(f"{base}/test/test_source2.tsv", sep="\t", dtype=str)
test_s3 = pd.read_csv(f"{base}/test/test_source3.tsv", sep="\t", dtype=str)

print("TRAIN")
print("source1:", train_s1.shape)
print("source2:", train_s2.shape)
print("source3:", train_s3.shape)
print("ground truth:", gt.shape)

print("\nTEST")
print("source1:", test_s1.shape)
print("source2:", test_s2.shape)
print("source3:", test_s3.shape)
```

```python
gt["matched_entity_ids"] = gt["matched_entity_ids"].fillna("")
gt["num_matches"] = gt["matched_entity_ids"].apply(lambda x: 0 if x.strip() == "" else len(x.split(",")))

print("Total S1 entities in ground truth:", len(gt))
print("Singletons (0 matches):", (gt["num_matches"] == 0).sum())
print("Exactly 1 match:", (gt["num_matches"] == 1).sum())
print("More than 1 match:", (gt["num_matches"] > 1).sum())
print("\nDistribution of match counts:")
print(gt["num_matches"].value_counts().sort_index())

all_matches = gt["matched_entity_ids"].apply(lambda x: x.split(",") if x.strip() else [])
flat = [m for sub in all_matches for m in sub if m.strip()]
s2_count = sum(1 for m in flat if m.startswith("S2-"))
s3_count = sum(1 for m in flat if m.startswith("S3-"))
print(f"\nTotal match links: {len(flat)}")
print(f"  From Source 2: {s2_count}")
print(f"  From Source 3: {s3_count}")
```

```python
sample_ids = gt[gt["num_matches"] > 0]["source1_entity_id"].sample(5, random_state=42).tolist()

for sid in sample_ids:
    row = gt[gt["source1_entity_id"] == sid].iloc[0]
    matches = row["matched_entity_ids"].split(",")
    
    print(f"\n{'='*80}")
    print(f"SOURCE1 [{sid}]:")
    s1_row = train_s1[train_s1["entity_id"] == sid].iloc[0]
    print(f"  name: {s1_row['business_name']}")
    print(f"  address: {s1_row['business_address']}")
    print(f"  country: {s1_row['country']}")
    
    for mid in matches:
        mid = mid.strip()
        src_df = train_s2 if mid.startswith("S2-") else train_s3
        m_row = src_df[src_df["entity_id"] == mid]
        if len(m_row) == 0:
            print(f"  [{mid}] NOT FOUND")
            continue
        m_row = m_row.iloc[0]
        print(f"  MATCH [{mid}]:")
        print(f"    name: {m_row['business_name']}")
        print(f"    address: {m_row['business_address']}")
        print(f"    country: {m_row['country']}")
```
